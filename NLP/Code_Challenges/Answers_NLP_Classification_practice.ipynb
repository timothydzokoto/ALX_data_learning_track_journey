## QN 1.

`X` is a **sparse matrix** where the text data has been converted into numerical form — specifically a **document-term matrix**.

Its shape is `(n_documents, n_unique_words)`, so in your example it's a `(2, 10)` matrix — 2 documents, 10 unique tokens.

Each row represents a document, each column represents a unique word (feature), and each cell value is the **word count** — how many times that word appears in that document.

Visualized, it looks roughly like this:

| | and | are | closely | fascinating | is | language | learning | linked | machine | natural | processing |
|---|---|---|---|---|---|---|---|---|---|---|---|
| doc 1 | 0 | 0 | 0 | 1 | 1 | 0 | 1 | 0 | 1 | 0 | 0 |
| doc 2 | 1 | 1 | 1 | 0 | 0 | 1 | 1 | 1 | 1 | 1 | 1 |

A few things worth knowing about `X`:

- It's a **scipy sparse matrix** (`csr_matrix`), not a regular numpy array. This is intentional — real-world text data has thousands of unique words but any single document uses very few of them, so most values are 0. Sparse format saves memory by only storing the non-zero values.
- You can convert it to a dense array with `X.toarray()` if you want to inspect it directly.
- It's what you'd pass into a model, e.g. `model.fit(X, labels)`.
- `feature_names` maps the column indices back to the actual words, so column 0 → `"and"`, column 1 → `"are"`, etc.
